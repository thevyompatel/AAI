### Part E: Challenge Exercise — Calculator Agent
Answers arithmetic queries (e.g. "Add 25 and 48", "Find square root of 225") using Python's `math` module and string processing — not the LLM.

```python
import math
import re

NUM = r"(-?\d+(?:\.\d+)?)"

def calculator_agent(query):
    query = query.lower().strip()

    match = re.match(rf"add {NUM} and {NUM}", query)
    if match:
        a, b = float(match.group(1)), float(match.group(2))
        return f"{a} + {b} = {a + b}"

    match = re.match(rf"subtract {NUM} and {NUM}", query)
    if match:
        a, b = float(match.group(1)), float(match.group(2))
        return f"{a} - {b} = {a - b}"

    match = re.match(rf"multiply {NUM} and {NUM}", query)
    if match:
        a, b = float(match.group(1)), float(match.group(2))
        return f"{a} * {b} = {a * b}"

    match = re.match(rf"divide {NUM} and {NUM}", query)
    if match:
        a, b = float(match.group(1)), float(match.group(2))
        if b == 0:
            return "Cannot divide by zero"
        return f"{a} / {b} = {a / b}"

    match = re.match(rf"(?:find )?square root of {NUM}", query)
    if match:
        n = float(match.group(1))
        if n < 0:
            return "Cannot find square root of a negative number"
        return f"Square root of {n} = {math.sqrt(n)}"

    return "Sorry, I can only add, subtract, multiply, divide, or find square roots. Try: 'Add 25 and 48'."


print("Calculator Agent")
print("Examples: 'Add 25 and 48', 'Multiply 15 and 8', 'Find square root of 225'")
print("Type 'exit' to quit.\n")

while True:
    query = input("You: ")
    if query.lower() == "exit":
        break
    print("Agent:", calculator_agent(query))
```

### Theory: Chatbot vs AI Agent

| Feature | Chatbot | AI Agent |
|---|---|---|
| Purpose | Conversational interaction | Completes tasks autonomously |
| Decision Making | Responds only to user input | Can plan and choose actions |
| Memory | Usually limited to conversation | Can maintain state and memory |
| Tool Usage | Rarely uses external tools | Can use APIs, databases, web search, calculators, files, etc. |
| Goal-Oriented | No | Yes |
| Autonomy | Low | High |
| Example | FAQ bot, customer support bot | Personal assistant, coding agent, research agent |

**Workflow — Chatbot:** User → LLM → Response

**Workflow — AI Agent:** User → Understand Goal → Reason → Use Tools (API, Calculator, Database, Web Search) → Observe Results → Generate Final Answer
